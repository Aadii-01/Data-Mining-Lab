3. Experiment 3: Finding 5 number summary for dataset and study of Box plot.
  Iris Dataset: Five-number summary, Box Plot
  Glass Identification Dataset

```python
import seaborn as sns

df = sns.load_dataset("iris")
```

Five number summary

```python
summary = df.describe()

print(summary.loc[
    ["min", "25%", "50%", "75%", "max"]
])
```

Individual feature summary

```python
for column in df.columns[:-1]:

    print("\n", column)

    print(df[column].describe())
```

Boxplot

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,6))

sns.boxplot(data=df.drop(columns="species"))

plt.title("Iris Dataset Box Plot")

plt.show()
```
