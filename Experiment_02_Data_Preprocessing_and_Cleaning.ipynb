````python
import pandas as pd
import numpy as np
````

````python
df = pd.read_excel("/content/Student_Preprocessing_Dataset_50_Records.xlsx")
````

````python
df.describe(include='all')
````

Check missing values

````python
df.isnull().sum()
````

Fill missing values

````python
df['Age'].fillna(df['Age'].mean(),inplace=True)
````

````python
df["Attendan"].fillna(df["Attendan"].mean(), inplace=True)
````

````python
df["Marks"].fillna(df["Marks"].mean(), inplace=True)
````

````python
df["Departme"].fillna(df["Departme"].mode()[0], inplace=True)
````

````python
df["City"].fillna(df["City"].mode()[0], inplace=True)
````

Remove duplicate records

````python
print(df.duplicated().sum())

df.drop_duplicates(inplace=True)
````

Data smoothing (Bin mean)

````python
df["Attendance_Bin"] = pd.cut(df["Attendan"], bins=5)

df["Attendance_Smoothed"] = (
    df.groupby("Attendance_Bin")["Attendan"]
      .transform("mean")
)

print(df[["Attendan","Attendance_Smoothed"]].head())
````

Label encoding

````python
df["Gender"] = pd.factorize(df["Gender"])[0]
df["Departme"] = pd.factorize(df["Departme"])[0]
df["City"] = pd.factorize(df["City"])[0]

df.head()
````

````python
print(df["Gender"].unique())
print(df["Departme"].unique())
print(df["City"].unique())
````

Min max normalization

```
xnorm = (x-xmin)/ (xmax-xmin)

```

````python
df["Attendance_Normalized"] = (
    (df["Attendan"] - df["Attendan"].min()) /
    (df["Attendan"].max() - df["Attendan"].min())
)

df["Marks_Normalized"] = (
    (df["Marks"] - df["Marks"].min()) /
    (df["Marks"].max() - df["Marks"].min())
)

df[["Attendan", "Attendance_Normalized",
    "Marks", "Marks_Normalized"]].head()
````

Z score standardization

```
Z =(X-u)/(alpha)
```

````python
df["Age_Standardized"] = (
    (df["Age"] - df["Age"].mean()) /
    df["Age"].std()
)

df["Attendance_Standardized"] = (
    (df["Attendan"] - df["Attendan"].mean()) /
    df["Attendan"].std()
)

df["Marks_Standardized"] = (
    (df["Marks"] - df["Marks"].mean()) /
    df["Marks"].std()
)

df[["Age", "Age_Standardized",
    "Attendan", "Attendance_Standardized",
    "Marks", "Marks_Standardized"]].head()
````
